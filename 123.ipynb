```python
from youtube_transcript_api import YouTubeTranscriptApi, TranscriptsDisabled, NoTranscriptFound, VideoUnavailable

def extract_video_id(url):
    if "watch?v=" in url:
        return url.split("watch?v=")[1].split("&")[0]
    elif "youtu.be/" in url:
        return url.split("youtu.be/")[1].split("?")[0]
    elif "shorts/" in url:
        return url.split("shorts/")[1].split("?")[0]
    else:
        raise ValueError("지원하지 않는 유튜브 URL 형식입니다.")

def get_full_transcript_from_url(url, languages=['ko','en']):
    video_id = extract_video_id(url)
    print("📌 video_id:", video_id)

    try:
        # 먼저 직접 자막 시도
        try:
            transcript = YouTubeTranscriptApi.get_transcript(video_id, languages=languages)
            print("✅ 정상 자막 불러오기 성공")
        except NoTranscriptFound:
            # 자동 생성 자막 시도
            transcript = YouTubeTranscriptApi.get_transcript(video_id, languages=languages, 
                                                             continue_after_error=True)
            print("⚡ 자동 생성 자막 불러오기 성공")

        full_text = " ".join([t['text'] for t in transcript])
        return full_text

    except TranscriptsDisabled:
        print("❌ 자막이 비활성화된 영상입니다.")
    except VideoUnavailable:
        print("❌ 영상이 존재하지 않거나 접근 불가입니다.")
    except Exception as e:
        print("❌ 자막 불러오기 실패:", e)

    return None

# ======================
# 🔥 테스트
# ======================
url = "https://youtube.com/shorts/e6soeg-eJ2c?si=rz8CiCpGBQ10LxhO"

text = get_full_transcript_from_url(url)

if text:
    print("\n📄 전체 자막:")
    print(text)
```
