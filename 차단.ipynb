```python
import os
import shutil
import subprocess

USER_PROFILE = r"C:\chrome_auto_profile"
CHROME_PATH = r"C:\Program Files\Google\Chrome\Application\chrome.exe"  # 크롬 설치 경로

# 1️⃣ 기존 프로필 삭제
if os.path.exists(USER_PROFILE):
    try:
        shutil.rmtree(USER_PROFILE)
        print("🗑 기존 프로필 삭제 완료")
    except Exception as e:
        print(f"❌ 프로필 삭제 실패: {e}")

# 2️⃣ 새 폴더 생성
try:
    os.makedirs(USER_PROFILE, exist_ok=True)
    print("📁 새 프로필 폴더 생성 완료")
except Exception as e:
    print(f"❌ 새 프로필 폴더 생성 실패: {e}")

# 3️⃣ 새 프로필로 크롬 실행 → 로그인 수행
try:
    subprocess.Popen([
        CHROME_PATH,
        f'--user-data-dir={USER_PROFILE}',
        '--start-maximized'
    ])
    print("🌐 새 프로필 크롬 실행 완료. 로그인 후 창을 닫으세요.")
except Exception as e:
    print(f"❌ 크롬 실행 실패: {e}")
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.action_chains import ActionChains
import time

URLS = [
    "https://www.youtube.com/shorts/e6soeg-eJ2c",
    "https://www.youtube.com/shorts/LwQRUFyDKNU"
]

USER_PROFILE = r"C:\chrome_auto_profile"

options = Options()
options.add_argument(f"--user-data-dir={USER_PROFILE}")
options.add_argument("--window-size=1280,800")
options.add_argument("--force-device-scale-factor=1")

driver = webdriver.Chrome(service=Service(), options=options)

try:
    wait = WebDriverWait(driver, 30)

    for URL in URLS:
        driver.get(URL)
        print(f"🌐 접속 중: {URL}")

        video_area = wait.until(
            EC.presence_of_element_located((By.TAG_NAME, "video"))
        )

        ActionChains(driver).move_to_element(video_area).perform()
        time.sleep(1)

        # =========================
        # 🔴 더보기 클릭 (빨간 점)
        # =========================
        offset_x = video_area.size['width'] - 50
        offset_y = 30

        driver.execute_script("""
        var video = arguments[0];
        var x = arguments[1];
        var y = arguments[2];

        var rect = video.getBoundingClientRect();
        var clickX = rect.left + x;
        var clickY = rect.top + y;

        var dot = document.createElement('div');
        dot.style.width = '15px';
        dot.style.height = '15px';
        dot.style.background = 'red';
        dot.style.borderRadius = '50%';
        dot.style.position = 'fixed';
        dot.style.left = clickX + 'px';
        dot.style.top = clickY + 'px';
        dot.style.zIndex = 9999;
        document.body.appendChild(dot);

        var el = document.elementFromPoint(clickX, clickY);
        if (el) { el.click(); }
        """, video_area, offset_x, offset_y)

        print("✅ 더보기 클릭 완료")
        time.sleep(1)

        # =========================
        # 🔵 채널 관심없음 클릭 (파란 점)
        # =========================
        offset_y = 220

        driver.execute_script("""
        var video = arguments[0];
        var x = arguments[1];
        var y = arguments[2];

        var rect = video.getBoundingClientRect();
        var clickX = rect.left + x;
        var clickY = rect.top + y;

        var dot = document.createElement('div');
        dot.style.width = '15px';
        dot.style.height = '15px';
        dot.style.background = 'blue';
        dot.style.borderRadius = '50%';
        dot.style.position = 'fixed';
        dot.style.left = clickX + 'px';
        dot.style.top = clickY + 'px';
        dot.style.zIndex = 9999;
        document.body.appendChild(dot);

        var el = document.elementFromPoint(clickX, clickY);
        if (el) { el.click(); }
        """, video_area, offset_x, offset_y)

        print("✅ 채널 관심없음 클릭 완료")
        time.sleep(2)

except Exception as e:
    print("❌ 오류 발생:", e)

finally:
    driver.quit()
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.action_chains import ActionChains
import time

URLS = [
    "https://www.youtube.com/shorts/e6soeg-eJ2c",
    "https://www.youtube.com/shorts/LwQRUFyDKNU"
]

USER_PROFILE = r"C:\chrome_auto_profile"

options = Options()
options.add_argument(f"--user-data-dir={USER_PROFILE}")
options.add_argument("--window-size=1280,800")
options.add_argument("--force-device-scale-factor=1")

# 🔥 브라우저 창 완전 숨김
options.add_argument("--headless=new")
options.add_argument("--disable-gpu")
options.add_argument("--no-sandbox")

driver = webdriver.Chrome(service=Service(), options=options)

try:
    wait = WebDriverWait(driver, 30)

    for URL in URLS:
        driver.get(URL)
        print(f"🌐 접속 중: {URL}")

        video_area = wait.until(
            EC.presence_of_element_located((By.TAG_NAME, "video"))
        )

        ActionChains(driver).move_to_element(video_area).perform()
        time.sleep(1)

        # =========================
        # 🔴 더보기 클릭 (점 제거)
        # =========================
        offset_x = video_area.size['width'] - 50
        offset_y = 30

        driver.execute_script("""
        var video = arguments[0];
        var x = arguments[1];
        var y = arguments[2];

        var rect = video.getBoundingClientRect();
        var clickX = rect.left + x;
        var clickY = rect.top + y;

        var el = document.elementFromPoint(clickX, clickY);
        if (el) { el.click(); }
        """, video_area, offset_x, offset_y)

        print("✅ 더보기 클릭 완료")
        time.sleep(1)

        # =========================
        # 🔵 채널 관심없음 클릭
        # =========================
        offset_y = 220

        driver.execute_script("""
        var video = arguments[0];
        var x = arguments[1];
        var y = arguments[2];

        var rect = video.getBoundingClientRect();
        var clickX = rect.left + x;
        var clickY = rect.top + y;

        var el = document.elementFromPoint(clickX, clickY);
        if (el) { el.click(); }
        """, video_area, offset_x, offset_y)

        print("✅ 채널 관심없음 클릭 완료")
        time.sleep(2)

except Exception as e:
    print("❌ 오류 발생:", e)

finally:
    driver.quit()
```
